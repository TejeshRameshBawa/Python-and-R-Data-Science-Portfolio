## Data Frame 1

```r
library(ggplot2)
library(tidyr)
```

```r
# Reshape the data from a wide format to a long format
data_melted_dataframe_1 <- dataframe_1 %>%
  pivot_longer(cols = -`Fraud Categories`,
               names_to = "Year",
               values_to = "Fraud Cases")

# View the reshaped data
print(data_melted_dataframe_1)
```

### Line Plot: Trends Over Time

```r
# Line plot with Year on x-axis and Fraud Cases on y-axis
ggplot(data_melted_dataframe_1, aes(x = Year, y = `Fraud Cases`, group = `Fraud Categories`, color = `Fraud Categories`)) +
  geom_line(size = 1) +
  geom_point(size = 2) +
  theme_minimal(base_size = 2) +
  theme(axis.text.x = element_text(angle = 90, hjust = 1, size = 8),
        axis.text.y = element_text(size=10),
        plot.title = element_text(size = 18, face = "bold"),
        axis.title.x = element_text(size = 14),
        axis.title.y = element_text(size = 14),
        legend.title = element_text(size = 12),
        legend.text = element_text(size = 8)
        ) +
  scale_y_continuous(
    breaks = seq(0, 600000, by = 50000)) +
  labs(title = "Fraud Trends by Category Over Time", x = "Year", y = "Fraud Cases", color = "Fraud Categories")
```

### Line Plot: Trends Over Time (Mean)

```r
# Calculating the mean for each year for all fraud categories
data_melted_summary <- data_melted_dataframe_1 %>%
  group_by(Year) %>%
  summarise(Mean = mean(`Fraud Cases`, na.rm = TRUE))

# Line chart to show central tendency mean over the years
ggplot(data_melted_summary, aes(x = Year, y = Mean, group = 1)) +
  geom_line(color = "blue", size = 1) +
  geom_point(color = "darkred", size = 2) +
  ggtitle("Trend of Mean Fraud Cases Over Time") +
  xlab("Year") +
  ylab("Mean Fraud Cases") +
  theme_minimal(base_size = 12) +
  theme(axis.text.x = element_text(angle = 45, hjust = 1, vjust = 1))
```

### Boxplot

```r
# Boxplot to visualise the dispersion across categories per year
ggplot(data_melted_dataframe_1, aes(x = `Fraud Categories`, y = `Fraud Cases`, fill = `Fraud Categories`)) +
  geom_boxplot(alpha = 0.6) +
  geom_jitter(width = 0.2, alpha = 0.5) +
  theme_minimal(base_size = 10) +
  theme(axis.text.x = element_text(angle = 90, hjust = 1, size = 10)) +
  scale_y_continuous(
    breaks = seq(0, 600000, by = 50000)) +
  labs(title = "Dispersion of Fraud Counts by Fraud Categories", x = "Fraud Categories", y = "Fraud Cases")
```

### Scatter Plot

```r
# visualise the scatter plot for fraud cases by year
ggplot(data_melted_dataframe_1, aes(x = Year, y = `Fraud Cases`, color = `Fraud Categories`)) +
  geom_point(size = 3, alpha = 1) +
  theme_minimal() +
  theme(
    axis.text.x = element_text(angle = 90, hjust = 1, size = 5)
  ) +
    scale_y_continuous(
    breaks = seq(0, 600000, by = 50000)) +
  labs(
    title = "Scatter Plot of Fraud Cases by Year",
    x = "Year",
    y = "Fraud Cases",
    color = "Fraud Categories"
  )
```

### Histogram

```r
# visualise a bar plot
ggplot(data_melted_dataframe_1, aes(x = `Fraud Categories`, y = `Fraud Cases`, fill = Year)) +
  geom_bar(stat = "identity", position = "dodge", alpha = 0.8) +
  theme_minimal(base_size = 10) +
  theme(
    axis.text.x = element_text(angle = 90, hjust = 1, size = 10),
  ) +
  scale_y_continuous(
    breaks = seq(0, 600000, by = 50000)) +
  labs(
    title = "Fraud Cases by Category and Year",
    x = "Fraud Categories",
    y = "Fraud Cases",
    fill = "Year"
  )
```

```r
# loading the dataset
file_path <- "/content/cw_r.xlsx"
data <- read_xlsx(file_path, sheet = "Table 5", skip=9)
```

```r
# renaming the columns for clarity
colnames(data) <- c("Area_Code", "Area_Name", "Number_of_Offences",
                    "Rate_per_1000_Population", "Change_from_Previous_Year")

# remove non-numeric rows from the Rate_per_1000_Population column
data <- data %>%
  mutate(Rate_per_1000_Population = as.numeric(gsub("[^0-9.]", "", Rate_per_1000_Population)))

# filtering and cleaning the data
clean_df <- data %>%
  filter(!is.na(Area_Name)) %>% # removing rows with any missing Area_Name
  filter(!is.na(Number_of_Offences) & !is.na(Rate_per_1000_Population)) %>% # removing rows with any missing data
  select(Area_Name, Number_of_Offences, Rate_per_1000_Population) # Keep all relevant columns

# converting columns to numeric
clean_df <- clean_df %>%
  mutate(
    Number_of_Offences = as.numeric(Number_of_Offences),
    Rate_per_1000_Population = as.numeric(Rate_per_1000_Population)
  ) %>%
  drop_na() # Rremove all rows with any invalid numeric values

print(clean_df)
```

```r
# preprocess the dataset
processed_data <- clean_df %>%
  mutate(
    Region = case_when(
      Area_Name %in% c("North East", "North West", "Yorkshire and The Humber",
                       "East Midlands", "West Midlands", "East",
                       "South East", "South West", "WALES") ~ Area_Name,
      TRUE ~ NA_character_
    )
  ) %>%
  fill(Region, .direction = "down") %>% # Assign all regions to counties
  filter(!Area_Name %in% c("ENGLAND AND WALES [note 13]", "ENGLAND", "London [note 14]")) %>% # Excluding the first 2 rows
  rename(County = Area_Name) %>% # renaming Area_Name to County
  select(Region, County, Number_of_Offences, Rate_per_1000_Population) # selecting relevant columns

print(processed_data)
```

```r
# summarize the data at the region level
region_summary <- processed_data %>%
  group_by(Region) %>%
  summarize(
    Total_Offences = sum(Number_of_Offences, na.rm = TRUE),
    Avg_Rate_per_1000 = mean(Rate_per_1000_Population, na.rm = TRUE)
  )

# Bar plot for Total Offences by Region
ggplot(region_summary, aes(x = reorder(Region, -Total_Offences), y = Total_Offences, fill = Region)) +
  geom_bar(stat = "identity", alpha = 0.8, color = "black") +
  theme_minimal() +
  labs(
    title = "Total Number of Offences by Region",
    x = "Region",
    y = "Total Offences"
  ) +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))

# Bar plot for Average Offence Rate per 1000 Population by Region
ggplot(region_summary, aes(x = reorder(Region, -Avg_Rate_per_1000), y = Avg_Rate_per_1000, fill = Region)) +
  geom_bar(stat = "identity", alpha = 0.8, color = "black") +
  theme_minimal() +
  labs(
    title = "Average Offence Rate per 1000 Population by Region",
    x = "Region",
    y = "Average Rate per 1000 Population"
  ) +
  theme(axis.text.x = element_text(angle = 45, hjust = 1))
```

```r
# summarize offences by county
county_summary <- processed_data %>%
  group_by(County) %>%
  summarize(Total_Offences = sum(Number_of_Offences, na.rm = TRUE)) %>%
  arrange(Total_Offences) # Sort by total offences in ascending order

# selecting the bottom counties
bottom_counties <- head(county_summary)

# visualize the bottom counties using a bar plot
ggplot(bottom_counties, aes(x = reorder(County, Total_Offences), y = Total_Offences, fill = County)) +
  geom_bar(stat = "identity", color = "black", alpha = 0.8) +
  theme_minimal() +
  labs(
    title = "Top Counties with the Lowest Total Offences",
    x = "County",
    y = "Total Offences"
  ) +
  theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
  scale_fill_brewer(palette = "Pastel1") # Optional color palette
```
